```python
# Spiral Art Generator using Turtle and Tkinter
# This version is fully commented to help beginners understand every part

import turtle  # Turtle is used for drawing on the canvas
import tkinter as tk  # Tkinter is used to create the GUI window and widgets
from tkinter import ttk  # ttk provides more styled widgets like Combobox
import time  # Time module helps us track delays or durations

# --- GLOBAL STATE VARIABLES ---

drawing = False        # Boolean flag to track if the spiral is currently drawing
spiral = None          # Will hold the turtle object used to draw the spiral
current_angle = 59     # Default turn angle for the spiral pattern
colors = []            # List to store the current set of colors for the spiral
loop_counter = 0       # Counter that tracks how many steps the turtle has taken

# --- COLOR THEMES ---
# A dictionary of color palette options. Each key is a name, and the value is a list of colors.
color_palettes = {
    "Rainbow": ["red", "orange", "yellow", "green", "blue", "indigo", "violet"],
    "Blue-Green": ["blue", "cyan", "green"],
    "Fire": ["yellow", "orange", "red", "darkred"],
    "Ocean": ["navy", "blue", "deepskyblue", "cyan"],
    "Pastel": ["plum", "lightpink", "lightblue", "lavender", "peachpuff"],
    "Black": ["black"]
}

# --- CHECK IF THE TURTLE HAS GONE OFF SCREEN ---
def is_turtle_off_screen():
    if spiral is None:
        return False
    x, y = spiral.pos()  # Get the current (x, y) position of the turtle
    width, height = 800, 800  # Dimensions of our canvas
    return abs(x) > width / 2 or abs(y) > height / 2  # If out of bounds, return True

# --- START OR CONTINUE THE SPIRAL DRAWING ---
def draw_spiral():
    global drawing, spiral, loop_counter
    if drawing:
        return  # If already drawing, exit function to avoid overlapping loops

    if spiral is None:
        setup_spiral()  # Initialize the turtle if it doesn't exist

    drawing = True  # Set flag to True so the while loop runs

    spiral.pensize(pen_size.get())  # Update pen size from slider value

    offscreen_time = None  # Used to track how long turtle has been off screen

    while drawing:
        spiral.color(colors[loop_counter % len(colors)])  # Cycle through colors
        spiral.forward(loop_counter)  # Move turtle forward
        spiral.left(current_angle)  # Turn turtle left
        loop_counter += 1  # Increment step counter

        if is_turtle_off_screen():  # Check if turtle left the screen
            if offscreen_time is None:
                offscreen_time = time.time()  # Record time of first exit
            elif time.time() - offscreen_time > 2:  # If gone > 2 seconds
                drawing = False  # Stop drawing
        else:
            offscreen_time = None  # Reset if turtle is back on screen

        root.update()  # Keep GUI responsive (relevant: search "tkinter root.update")

# --- INITIALIZE THE TURTLE OBJECT ---
def setup_spiral():
    global spiral
    spiral = turtle.RawTurtle(canvas)  # Attach turtle to our canvas
    spiral.speed(10)  # Set speed to fast (search "turtle speed")
    spiral.hideturtle()  # Hide turtle icon for a cleaner look
    spiral.penup()  # Lift pen so it doesn't draw while moving to center
    spiral.goto(0, 0)  # Move turtle to the center of the screen
    spiral.setheading(0)  # Reset direction to right (0 degrees)
    spiral.pendown()  # Put pen down to start drawing

# --- STOP THE DRAWING WITHOUT RESETTING ANYTHING ---
def stop_drawing():
    global drawing
    drawing = False  # Stop the while loop in draw_spiral()

# --- CLEAR THE SCREEN AND RESET EVERYTHING ---
def reset_all():
    global drawing, spiral, loop_counter
    drawing = False  # Stop current drawing
    loop_counter = 0  # Reset step counter
    canvas.delete("all")  # Clear the canvas
    screen.reset()  # Reset turtle screen state
    screen.bgcolor("white")  # Set background to white
    spiral = None  # Delete the spiral turtle so we can recreate it later

# --- APPLY THE CURRENT PEN SIZE TO THE TURTLE ---
def apply_pen_size():
    if spiral is not None:
        spiral.pensize(pen_size.get())  # Set pen size to slider value

# --- APPLY THE CURRENT ANGLE TO CHANGE SPIRAL TIGHTNESS ---
def apply_angle():
    global current_angle
    current_angle = angle.get()  # Update the angle value from slider

# --- CHANGE THE COLOR PALETTE BASED ON DROPDOWN ---
def apply_color_style():
    global colors
    selected = color_choice.get()  # Get selected palette name from dropdown
    colors = color_palettes.get(selected, ["black"])  # Get the corresponding color list

# --- GUI SETUP (TKINTER WINDOW) ---
root = tk.Tk()  # Create the main GUI window (search "tkinter root")
root.title("Spiral Art Generator")  # Set the window title

# Create frame on the left side for buttons and sliders
control_frame = tk.Frame(root)
control_frame.pack(side=tk.LEFT, padx=10, pady=10)

# Create frame for turtle canvas on the right side
canvas_frame = tk.Frame(root)
canvas_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)

# Create a scrollable canvas and turtle screen
canvas = turtle.ScrolledCanvas(canvas_frame, width=800, height=800)  # Wider canvas
canvas.pack(fill=tk.BOTH, expand=True)
screen = turtle.TurtleScreen(canvas)  # Convert canvas into turtle screen
screen.bgcolor("white")  # Set initial background color

# --- CREATE GUI SLIDERS AND BUTTONS ---

# Angle control slider
tk.Label(control_frame, text="Turn Angle").pack()
angle = tk.Scale(control_frame, from_=1, to=179, orient=tk.HORIZONTAL)
angle.set(current_angle)  # Set initial value
angle.pack()
tk.Button(control_frame, text="Apply Angle", command=apply_angle).pack(pady=2)

# Pen size slider
tk.Label(control_frame, text="Pen Size").pack()
pen_size = tk.Scale(control_frame, from_=1, to=10, orient=tk.HORIZONTAL)
pen_size.set(2)  # Set initial value
pen_size.pack()
tk.Button(control_frame, text="Apply Pen Size", command=apply_pen_size).pack(pady=2)

# Color palette dropdown
tk.Label(control_frame, text="Color Style").pack()
color_choice = ttk.Combobox(control_frame, values=list(color_palettes.keys()))  # Dropdown with color themes
color_choice.set("Rainbow")  # Default selection
color_choice.pack()
tk.Button(control_frame, text="Apply Color Style", command=apply_color_style).pack(pady=2)

# Action buttons
tk.Button(control_frame, text="Draw Spiral", command=draw_spiral).pack(pady=5)  # Start drawing
tk.Button(control_frame, text="Stop", command=stop_drawing).pack(pady=5)        # Pause drawing
tk.Button(control_frame, text="Reset", command=reset_all).pack(pady=5)          # Reset everything
tk.Button(control_frame, text="Exit", command=root.destroy).pack(pady=5)        # Close window

# Initialize color palette before first drawing
apply_color_style()

# Run the main event loop to show the GUI
root.mainloop()  # Keeps the GUI running and waits for user interaction

```
