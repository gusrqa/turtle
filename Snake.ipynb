```python
# 🐍 Snake Game in Turtle for Beginners
# This version includes:
# - Arrow key controls
# - Screen wrap-around (snake reappears on opposite edge)
# - Game over message without overlapping score
# - Reduced screen size (400x400)
# - Full beginner-friendly comments

import turtle      # Turtle module is used for drawing shapes and screen control
import time        # Time module helps us pause the game to control speed
import random      # Random module is used to place the food in random positions

# ------------------ SCREEN SETUP ------------------

win = turtle.Screen()                       # Create the main game window
win.title("Snake Game")                     # Set the window title
win.bgcolor("black")                        # Set background color to black
win.setup(width=400, height=400)            # Set screen size to 400x400 pixels
win.tracer(0)                               # Disable auto-refresh for smoother animation

# ------------------ SNAKE HEAD SETUP ------------------

head = turtle.Turtle()                      # Create the snake's head
head.speed(0)                               # Max drawing speed (0 is fastest)
head.shape("square")                        # Use square shape
head.color("green")                         # Color the head green
head.penup()                                # Prevent turtle from drawing lines
head.goto(0, 0)                             # Start at the center of the screen
head.direction = "stop"                     # Initial state is stopped (no movement)

# ------------------ FOOD SETUP ------------------

food = turtle.Turtle()                      # Create the food
food.speed(0)                               # Instant drawing
food.shape("circle")                        # Round food
food.color("red")                           # Red food
food.penup()                                # No drawing line
food.goto(0, 100)                           # Place food a bit above the snake

# ------------------ SCORE AND TEXT SETUP ------------------

segments = []                               # List to store body segments
score = 0                                   # Start score
high_score = 0                              # Best score saved in memory

# Pen for writing score
pen = turtle.Turtle()                       # Create turtle for score text
pen.speed(0)                                # Instant display
pen.color("white")                          # White text
pen.penup()                                 # Don’t draw line
pen.hideturtle()                            # Hide the turtle shape
pen.goto(0, 170)                            # Top-center of screen
pen.write("Press arrow key to start", align="center", font=("Courier", 14, "normal"))

# Message pen (for Game Over)
message = turtle.Turtle()                   # Another turtle for text messages
message.speed(0)
message.color("white")
message.penup()
message.hideturtle()
message.goto(0, 0)                          # Center of the screen

# ------------------ GAME STATE ------------------

running = False                             # Boolean: True = running, False = paused/game over

# ------------------ MOVEMENT FUNCTIONS ------------------

# This function starts the game and sets the direction
def start_and_set_dir(direction):
    global running
    if not running:
        running = True                      # Game starts
        pen.clear()                         # Clear "Press to start"
        message.clear()                     # Clear any "Game Over"
        update_score()                      # Write initial score
    head.direction = direction              # Set the movement direction

# Arrow key handlers
def go_up(): start_and_set_dir("up")
def go_down(): start_and_set_dir("down")
def go_left(): start_and_set_dir("left")
def go_right(): start_and_set_dir("right")

# Move the head forward in the current direction
def move():
    if head.direction == "up":
        head.sety(head.ycor() + 20)         # Move up by 20 units
    elif head.direction == "down":
        head.sety(head.ycor() - 20)         # Move down
    elif head.direction == "left":
        head.setx(head.xcor() - 20)         # Move left
    elif head.direction == "right":
        head.setx(head.xcor() + 20)         # Move right

    # Edge wrapping logic
    if head.xcor() > 190:                   # If snake crosses right edge
        head.setx(-190)                     # Wrap to left
    elif head.xcor() < -190:                # If snake crosses left edge
        head.setx(190)                      # Wrap to right
    elif head.ycor() > 190:                 # If snake crosses top
        head.sety(-190)                     # Wrap to bottom
    elif head.ycor() < -190:                # If snake crosses bottom
        head.sety(190)                      # Wrap to top

# ------------------ KEYBOARD BINDINGS ------------------

win.listen()                                # Allow keyboard input
win.onkeypress(go_up, "Up")                 # Arrow Up
win.onkeypress(go_down, "Down")             # Arrow Down
win.onkeypress(go_left, "Left")             # Arrow Left
win.onkeypress(go_right, "Right")           # Arrow Right

# ------------------ SCORE DISPLAY FUNCTION ------------------

def update_score():
    pen.clear()                             # Clear previous score text
    pen.write(f"Score: {score}  High Score: {high_score}",
              align="center", font=("Courier", 14, "normal"))

# ------------------ MAIN GAME LOOP ------------------

delay = 0.1                                 # Controls game speed (lower = faster)

while True:
    win.update()                            # Manually refresh screen

    if running:                             # Run only if game is active
        # --- FOOD COLLISION CHECK ---
        if head.distance(food) < 20:        # If snake is close to food
            # Move food to a random location
            x = random.randint(-180, 180)
            y = random.randint(-180, 180)
            food.goto(x, y)

            # Create a new segment for the snake body
            new_segment = turtle.Turtle()
            new_segment.speed(0)
            new_segment.shape("square")
            new_segment.color("lightgreen")
            new_segment.penup()
            segments.append(new_segment)    # Add to the list of segments

            # Update score
            score += 10
            if score > high_score:
                high_score = score          # Update high score if necessary
            update_score()

        # --- MOVE BODY SEGMENTS ---
        # Start from the end and follow the one in front
        for i in range(len(segments) - 1, 0, -1):
            x = segments[i - 1].xcor()
            y = segments[i - 1].ycor()
            segments[i].goto(x, y)

        # First segment follows the head
        if segments:
            segments[0].goto(head.xcor(), head.ycor())

        move()                              # Move the head

        # --- BODY COLLISION CHECK ---
        for segment in segments:
            if segment.distance(head) < 20:   # If snake hits itself
                time.sleep(1)
                head.goto(0, 0)               # Return to center
                head.direction = "stop"       # Stop moving

                for segment in segments:
                    segment.goto(1000, 1000)  # Move body segments off-screen
                segments.clear()              # Empty segment list
                score = 0                     # Reset score
                update_score()
                message.clear()
                message.write("Game Over! Press arrow key to restart",
                              align="center", font=("Courier", 14, "normal"))
                running = False               # Pause the game

    time.sleep(delay)                         # Pause before next frame (controls speed)

```
