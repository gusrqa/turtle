```python
# ---------------------------------------------------------------
# 🏓 PONG GAME - FULLY COMMENTED FOR BEGINNERS
# ---------------------------------------------------------------
# This game shows how to build a simple two-player Pong game in Python.
# It uses the built-in 'turtle' module for drawing and animation.
# Every line has a comment to explain what it does, step by step!
# ---------------------------------------------------------------

import turtle  # We import the turtle module to draw shapes and make things move

# 🖼️ Set up the game window (the screen)
win = turtle.Screen()  # Create a window for the game
win.title("Simple Pong with Rainbow Ball!")  # Set the title of the window
win.bgcolor("black")  # Set the background color to black
win.setup(width=800, height=600)  # Set the size of the window (800 wide, 600 tall)
win.tracer(0)  # Turn off automatic animation updates for better performance

# 🧮 Start the game with zero points for both players
score_a = 0  # Player A starts with 0 points
score_b = 0  # Player B starts with 0 points

# 🌈 List of colors that the ball will cycle through like a rainbow
rainbow_colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
color_index = 0  # This tells us which color to show next

ball_moving = False  # At the beginning, the ball does not move until we press space

# 🏓 Create the left paddle (Player A)
paddle_a = turtle.Turtle()  # Create a new turtle object for paddle A
paddle_a.speed(0)  # Speed of drawing (0 is the fastest)
paddle_a.shape("square")  # Shape of the paddle
paddle_a.color("white")  # Paddle color
paddle_a.shapesize(stretch_wid=6, stretch_len=1)  # Make the square tall like a paddle
paddle_a.penup()  # Don't draw a line when moving
paddle_a.goto(-350, 0)  # Place paddle A on the left side

# 🏓 Create the right paddle (Player B)
paddle_b = turtle.Turtle()  # Create a new turtle object for paddle B
paddle_b.speed(0)
paddle_b.shape("square")
paddle_b.color("white")
paddle_b.shapesize(stretch_wid=6, stretch_len=1)
paddle_b.penup()
paddle_b.goto(350, 0)  # Place paddle B on the right side

# 🟠 Create the ball
ball = turtle.Turtle()  # Create a turtle object for the ball
ball.speed(0)  # Fastest animation
ball.shape("circle")  # Ball shape is a circle
ball.color(rainbow_colors[color_index])  # Start with the first color in the rainbow
ball.penup()  # Don't leave a trail when moving
ball.goto(0, 0)  # Start in the center of the screen
ball.shapesize(stretch_wid=1, stretch_len=1)  # Default size
ball.dx = 0.2  # Change in x position (speed right or left)
ball.dy = 0.2  # Change in y position (speed up or down)

# 🧾 Create the score display
pen = turtle.Turtle()  # Turtle object to draw text
pen.speed(0)
pen.color("white")
pen.penup()
pen.hideturtle()  # We don't need to see this turtle
pen.goto(0, 260)  # Place the text at the top center
pen.write("Player A: 0  Player B: 0", align="center", font=("Courier", 24, "normal"))

# 📝 Message that tells players to press SPACE to start
prompt = turtle.Turtle()
prompt.speed(0)
prompt.color("white")
prompt.penup()
prompt.hideturtle()
prompt.goto(0, -270)  # Place message near the bottom of the screen

# This function shows the message to press SPACE
def show_prompt():
    prompt.clear()
    prompt.write("Press SPACE to serve", align="center", font=("Courier", 18, "italic"))

# This function hides the message
def hide_prompt():
    prompt.clear()

# 🎮 Move the left paddle up
def paddle_a_up():
    y = paddle_a.ycor()  # Get current y position
    if y < 250:  # Don't go too far up
        y += 20  # Move up by 20
    paddle_a.sety(y)  # Set new position

# 🎮 Move the left paddle down
def paddle_a_down():
    y = paddle_a.ycor()
    if y > -250:  # Don't go too far down
        y -= 20
    paddle_a.sety(y)

# 🎮 Move the right paddle up
def paddle_b_up():
    y = paddle_b.ycor()
    if y < 250:
        y += 20
    paddle_b.sety(y)

# 🎮 Move the right paddle down
def paddle_b_down():
    y = paddle_b.ycor()
    if y > -250:
        y -= 20
    paddle_b.sety(y)

# 🔄 Update the score display
def update_score():
    pen.clear()  # Erase old text
    pen.write(f"Player A: {score_a}  Player B: {score_b}", align="center", font=("Courier", 24, "normal"))

# 🛑 Reset everything if someone wins
def reset_game():
    global score_a, score_b
    score_a = 0  # Reset both scores
    score_b = 0
    update_score()  # Show new scores
    reset_ball()  # Start over

# 🟢 Put the ball back in the center and wait to serve
def reset_ball():
    global ball_moving
    ball.goto(0, 0)  # Center the ball
    ball.dx = 0.2 if ball.dx > 0 else -0.2  # Keep direction
    ball.dy = 0.2
    ball_moving = False  # Ball stops moving
    show_prompt()  # Show the message

# ▶️ Start the ball when spacebar is pressed
def start_ball():
    global ball_moving
    if not ball_moving:
        ball_moving = True
        hide_prompt()

# 🎹 Connect the keyboard to the paddle movement
win.listen()
win.onkeypress(paddle_a_up, "w")
win.onkeypress(paddle_a_down, "s")
win.onkeypress(paddle_b_up, "Up")
win.onkeypress(paddle_b_down, "Down")
win.onkeypress(start_ball, "space")

# 👋 Show the first "Press SPACE" message
show_prompt()

# 🔁 This is the main game loop — it runs forever
while True:
    win.update()  # Refresh the screen

    # Only move the ball if the game is active
    if ball_moving:
        # Move the ball by its speed
        ball.setx(ball.xcor() + ball.dx)
        ball.sety(ball.ycor() + ball.dy)

        # Cycle the ball's color
        color_index = (color_index + 1) % len(rainbow_colors)
        ball.color(rainbow_colors[color_index])

        # Bounce off the top wall
        if ball.ycor() > 290:
            ball.sety(290)
            ball.dy *= -1  # Change direction

        # Bounce off the bottom wall
        if ball.ycor() < -290:
            ball.sety(-290)
            ball.dy *= -1

        # If ball hits right wall, Player A scores
        if ball.xcor() > 390:
            score_a += 1
            update_score()
            reset_ball()

        # If ball hits left wall, Player B scores
        if ball.xcor() < -390:
            score_b += 1
            update_score()
            reset_ball()

        # If ball hits paddle B (right paddle)
        if (340 < ball.xcor() < 350) and (paddle_b.ycor() - 50 < ball.ycor() < paddle_b.ycor() + 50):
            ball.setx(340)
            ball.dx *= -1  # Bounce back

        # If ball hits paddle A (left paddle)
        if (-350 < ball.xcor() < -340) and (paddle_a.ycor() - 50 < ball.ycor() < paddle_a.ycor() + 50):
            ball.setx(-340)
            ball.dx *= -1

        # Check if someone won
        if score_a == 5:
            pen.goto(0, 0)
            pen.write("Player A Wins!", align="center", font=("Courier", 36, "bold"))
            win.update()
            turtle.time.sleep(2)
            pen.goto(0, 260)
            reset_game()

        if score_b == 5:
            pen.goto(0, 0)
            pen.write("Player B Wins!", align="center", font=("Courier", 36, "bold"))
            win.update()
            turtle.time.sleep(2)
            pen.goto(0, 260)
            reset_game()
```
