```python
# ---------------------------------------------------------------
# 🎮 SIMPLE PONG GAME IN PYTHON (USING turtle MODULE)
# ---------------------------------------------------------------
#
# 🔍 ABOUT THIS PROJECT:
# This is a beginner-friendly Pong game written in Python.
# It's built using the standard `turtle` graphics module — no extra libraries needed!
# Two players can control paddles and try to bounce the ball past each other.
#
# 🧠 WHAT YOU'LL LEARN:
# - How to use the turtle module for graphics and drawing
# - How to use loops, variables, and conditionals in Python
# - How to create game logic and basic physics (collisions, movement)
# - How to capture user input using keyboard events
# - How to manage game state, display scores, and respond to game events
#
# 🕹️ HOW TO PLAY:
# - Player A (left paddle): Use the **W** key (up) and **S** key (down)
# - Player B (right paddle): Use the **Up Arrow** and **Down Arrow**
# - Press the **Spacebar** to serve the ball after each point
# - First player to 5 points wins. Game resets automatically.
#
# 🌈 COOL FEATURES INCLUDED:
# - Rainbow-colored ball that changes every frame
# - Paddle and ball customization (size, color, speed)
# - "Press SPACE to serve" prompt for pacing the game
#
# ✏️ HOW TO CUSTOMIZE:
# - Change paddle colors or sizes: Look for paddle_a / paddle_b section
# - Make the ball faster or larger: See `ball.dx`, `ball.shapesize()`
# - Change the background color: `win.bgcolor(...)`
# - Adjust the winning score or scoring messages
#
# 🧱 EXTENSIONS YOU CAN BUILD:
# - Add sound effects or background music
# - Add countdown timer before serve
# - Add player name input and display
# - Add single-player mode with AI paddle
# - Add score history and difficulty settings
#
# 👶 LEVEL: Beginner to Intermediate Python
# ✅ Python Version: 3.6 or higher
# 🚀 Let's build, learn, and have fun!
#
# ---------------------------------------------------------------

import turtle

# ---------------------- WINDOW SETUP ------------------------
# Create game window
win = turtle.Screen()
win.title("Simple Pong with Rainbow Ball!")
win.bgcolor("black")  # You can change background color here
win.setup(width=800, height=600)
win.tracer(0)  # Turns off automatic screen updates for smoother gameplay

# ---------------------- GAME STATE VARIABLES ------------------------
score_a = 0
score_b = 0

# Colors for the rainbow effect on the ball
rainbow_colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
color_index = 0  # Controls current color in the rainbow cycle

ball_moving = False  # Controls whether the ball is in motion

# ---------------------- PADDLE A (LEFT PLAYER) ------------------------
paddle_a = turtle.Turtle()
paddle_a.speed(0)  # Animation speed (0 is the fastest)
paddle_a.shape("square")
paddle_a.color("white")  # <== CHANGE paddle A color here (e.g., "blue", "green", "#ff00ff")
paddle_a.shapesize(stretch_wid=6, stretch_len=1)  # <== CHANGE paddle height/width here
# For example: stretch_wid=8 makes it taller; stretch_len=2 makes it thicker
paddle_a.penup()
paddle_a.goto(-350, 0)

# ---------------------- PADDLE B (RIGHT PLAYER) ------------------------
paddle_b = turtle.Turtle()
paddle_b.speed(0)
paddle_b.shape("square")
paddle_b.color("white")  # <== CHANGE paddle B color here
paddle_b.shapesize(stretch_wid=6, stretch_len=1)  # <== CHANGE paddle size here
paddle_b.penup()
paddle_b.goto(350, 0)

# ---------------------- BALL ------------------------
ball = turtle.Turtle()
ball.speed(0)
ball.shape("circle")
ball.color(rainbow_colors[color_index])  # Starts with first color in the rainbow
ball.penup()
ball.goto(0, 0)

# BALL SIZE:
# To make the ball bigger, use shapesize(). Default is 1.
# Example: ball.shapesize(stretch_wid=2, stretch_len=2) makes it twice as big
ball.shapesize(stretch_wid=1, stretch_len=1)  # <== CHANGE ball size here

# BALL SPEED:
ball.dx = 0.2  # <== CHANGE horizontal speed here (increase for faster ball)
ball.dy = 0.2  # <== CHANGE vertical speed here

# ---------------------- SCORE DISPLAY ------------------------
pen = turtle.Turtle()
pen.speed(0)
pen.color("white")
pen.penup()
pen.hideturtle()
pen.goto(0, 260)
pen.write("Player A: 0  Player B: 0", align="center", font=("Courier", 24, "normal"))

# ---------------------- SPACEBAR PROMPT ------------------------
prompt = turtle.Turtle()
prompt.speed(0)
prompt.color("white")
prompt.penup()
prompt.hideturtle()
prompt.goto(0, -270)

def show_prompt():
    prompt.clear()
    prompt.write("Press SPACE to serve", align="center", font=("Courier", 18, "italic"))

def hide_prompt():
    prompt.clear()

# ---------------------- PADDLE MOVEMENT FUNCTIONS ------------------------
def paddle_a_up():
    y = paddle_a.ycor()
    if y < 250:
        y += 20  # <== Increase for faster paddle movement
    paddle_a.sety(y)

def paddle_a_down():
    y = paddle_a.ycor()
    if y > -250:
        y -= 20
    paddle_a.sety(y)

def paddle_b_up():
    y = paddle_b.ycor()
    if y < 250:
        y += 20
    paddle_b.sety(y)

def paddle_b_down():
    y = paddle_b.ycor()
    if y > -250:
        y -= 20
    paddle_b.sety(y)

# ---------------------- SCORE AND GAME RESET ------------------------
def update_score():
    pen.clear()
    pen.write(f"Player A: {score_a}  Player B: {score_b}", align="center", font=("Courier", 24, "normal"))

def reset_game():
    global score_a, score_b
    score_a = 0
    score_b = 0
    update_score()
    reset_ball()

def reset_ball():
    global ball_moving
    ball.goto(0, 0)
    # Make sure the ball continues in the same horizontal direction it had
    ball.dx = 0.2 if ball.dx > 0 else -0.2
    ball.dy = 0.2
    ball_moving = False
    show_prompt()

def start_ball():
    global ball_moving
    if not ball_moving:
        ball_moving = True
        hide_prompt()

# ---------------------- KEYBOARD CONTROLS ------------------------
win.listen()
win.onkeypress(paddle_a_up, "w")        # Player A: Move up
win.onkeypress(paddle_a_down, "s")      # Player A: Move down
win.onkeypress(paddle_b_up, "Up")       # Player B: Move up
win.onkeypress(paddle_b_down, "Down")   # Player B: Move down
win.onkeypress(start_ball, "space")     # Start serve

# Show prompt before first round
show_prompt()

# ---------------------- MAIN GAME LOOP ------------------------
while True:
    win.update()

    if ball_moving:
        # Move the ball
        ball.setx(ball.xcor() + ball.dx)
        ball.sety(ball.ycor() + ball.dy)

        # Cycle the ball's color each frame
        color_index = (color_index + 1) % len(rainbow_colors)
        ball.color(rainbow_colors[color_index])

        # Bounce off top and bottom borders
        if ball.ycor() > 290:
            ball.sety(290)
            ball.dy *= -1

        if ball.ycor() < -290:
            ball.sety(-290)
            ball.dy *= -1

        # Ball hits right wall → point for Player A
        if ball.xcor() > 390:
            score_a += 1
            update_score()
            reset_ball()

        # Ball hits left wall → point for Player B
        if ball.xcor() < -390:
            score_b += 1
            update_score()
            reset_ball()

        # Paddle B collision
        if (340 < ball.xcor() < 350) and (paddle_b.ycor() - 50 < ball.ycor() < paddle_b.ycor() + 50):
            ball.setx(340)
            ball.dx *= -1

        # Paddle A collision
        if (-350 < ball.xcor() < -340) and (paddle_a.ycor() - 50 < ball.ycor() < paddle_a.ycor() + 50):
            ball.setx(-340)
            ball.dx *= -1

        # Winning condition
        if score_a == 5:
            pen.goto(0, 0)
            pen.write("Player A Wins!", align="center", font=("Courier", 36, "bold"))
            win.update()
            turtle.time.sleep(2)
            pen.goto(0, 260)
            reset_game()

        if score_b == 5:
            pen.goto(0, 0)
            pen.write("Player B Wins!", align="center", font=("Courier", 36, "bold"))
            win.update()
            turtle.time.sleep(2)
            pen.goto(0, 260)
            reset_game()
```
