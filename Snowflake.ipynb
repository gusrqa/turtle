```python
# Import the turtle module to draw shapes
import turtle

# Import tkinter for creating the graphical user interface (GUI)
import tkinter as tk

# Import ttk for more advanced GUI widgets (like dropdowns)
from tkinter import ttk

# Import math for calculating angles and trigonometric functions
import math

# ------------------ Color List ------------------

# This is a list of all the color names turtle understands and can use
turtle_colors = [
    'black', 'white', 'gray', 'red', 'blue', 'green', 'yellow',
    'cyan', 'magenta', 'orange', 'purple', 'brown', 'pink',
    'lightblue', 'lightgreen', 'gold', 'darkgreen', 'navy'
]

# ------------------ FRACTAL DRAWING FUNCTION ------------------

# This function draws a Koch curve, which is the base shape of the snowflake
# 'length' is how long each side is
# 'level' is the depth of recursion, which controls how detailed the shape is
# 'alternate' is a trick to make it look more interesting (flips the angles)
def draw_koch(length, level, alternate=False):
    if level == 0:
        pen.forward(length)  # Base case: just draw a straight line
    else:
        length /= 3.0  # Divide the line into three smaller segments
        draw_koch(length, level - 1, alternate)  # First segment
        pen.left(60 if not alternate or level % 2 == 1 else -60)  # Turn left
        draw_koch(length, level - 1, alternate)  # Second segment (upward spike)
        pen.right(120 if not alternate or level % 2 == 1 else -120)  # Turn right
        draw_koch(length, level - 1, alternate)  # Third segment (downward spike)
        pen.left(60 if not alternate or level % 2 == 1 else -60)  # Turn back
        draw_koch(length, level - 1, alternate)  # Final segment

# This function draws the entire snowflake by repeating the Koch shape in a circle
def draw_snowflake():
    pen.clear()  # Clear the previous drawing

    # Get the number of sides/arms and recursion depth from sliders
    sides = sides_slider.get()
    depth = depth_slider.get()

    # These set the drawing canvas size and margin
    canvas_width, canvas_height = 1000, 1000
    margin = 100  # Space around the snowflake

    # Calculate the radius that fits the snowflake within the canvas
    fit_radius = (min(canvas_width, canvas_height) // 2) - margin

    # Adjust size based on how deep the recursion goes (complexity makes it bigger)
    scale_factor = (4 / 3) ** depth  # Koch curves grow with each recursion
    angle = math.pi / sides  # Calculate angle for even rotation
    side_length = 2 * fit_radius * math.sin(angle) / scale_factor  # Size of each arm
    radius = side_length / (2 * math.sin(angle))  # Radius of the full shape

    # Move turtle to center of the screen
    pen.penup()
    pen.home()
    pen.setheading(90)  # Point turtle upwards
    pen.forward(radius)  # Move out to start drawing from edge
    pen.setheading(0)  # Face right
    pen.pendown()

    # Set turtle color based on dropdown selection
    selected_color = color_choice.get()
    if selected_color not in turtle_colors:
        selected_color = "cyan"  # fallback color
    pen.color(selected_color)

    # Draw the fractal snowflake in a full circle
    for _ in range(sides):
        draw_koch(side_length, depth, alternate=True)  # Draw one arm
        pen.right(360 / sides)  # Rotate turtle evenly for the next arm

    screen.update()  # Refresh the turtle screen after drawing

# Clears the turtle canvas
def clear_screen():
    pen.clear()
    screen.update()

# ------------------ GUI SETUP ------------------

# Create the main application window
root = tk.Tk()
root.title("Fractal Snowflake Generator")  # Window title

# This frame holds the controls on the left
control_frame = tk.Frame(root, padx=10, pady=10, bg="lightgray")
control_frame.pack(side=tk.LEFT, fill=tk.Y)

# This frame holds the turtle canvas on the right
canvas_frame = tk.Frame(root)
canvas_frame.pack(side=tk.RIGHT)

# Set the size of the turtle drawing area
canvas_width, canvas_height = 1000, 1000

# Create a scrollable canvas using turtle inside the canvas frame
canvas = turtle.ScrolledCanvas(canvas_frame, width=canvas_width, height=canvas_height)
canvas.pack()

# Create a turtle screen inside the scrollable canvas
screen = turtle.TurtleScreen(canvas)
screen.bgcolor("black")  # Set background color
screen.tracer(0, 0)  # Disable animation for instant drawing

# Create the actual turtle used for drawing
pen = turtle.RawTurtle(screen)
pen.hideturtle()  # Hide turtle cursor
pen.speed(0)  # Fastest drawing speed

# ------------------ GUI WIDGETS ------------------

# Add a label and slider for number of arms/sides
tk.Label(control_frame, text="❄️ Number of Arms", bg="lightgray").pack()
sides_slider = tk.Scale(control_frame, from_=3, to=25, orient=tk.HORIZONTAL)
sides_slider.set(6)  # Default value
sides_slider.pack(pady=5)

# Add a label and slider for recursion depth
tk.Label(control_frame, text="🔁 Recursion Depth", bg="lightgray").pack()
depth_slider = tk.Scale(control_frame, from_=0, to=6, orient=tk.HORIZONTAL)  # 0 to 6 only
depth_slider.set(2)
depth_slider.pack(pady=5)

# Add a dropdown menu to choose the color
tk.Label(control_frame, text="🎨 Snowflake Color", bg="lightgray").pack(pady=(10, 0))
color_choice = ttk.Combobox(control_frame, values=turtle_colors, state="readonly")
color_choice.set("cyan")  # Default color
color_choice.pack(pady=5)

# Add a button to draw the snowflake
preview_btn = tk.Button(control_frame, text="👁 Preview", command=draw_snowflake)
preview_btn.pack(pady=10)

# Add a button to clear the screen
clear_btn = tk.Button(control_frame, text="🧹 Clear", command=clear_screen)
clear_btn.pack(pady=5)

# ------------------ INITIAL DRAW ------------------

# Draw one snowflake right when the program starts
draw_snowflake()

# Start the GUI event loop
root.mainloop()
```
